```python
from typing import Callable

def empty_deco(func: Callable):
    def wrapper():
        res = func()
        return res
    return wrapper
```

```python
@empty_deco
def my_func():
    return 124
```

```python
print(my_func())
```

## Декоратор для подсчета времени

```python
from typing import Callable
import time

def timer_deco(func: Callable):
    def wrapper():
        start = time.time()
        res = func()
        end = time.time()
        print(f"Исполнение заняло {end-start}")
        return res
    return wrapper
```

```python
@timer_deco
def my_func():
    time.sleep(1.5)
    return 124

print(my_func())
```

## Функция с параметром

```python
from typing import Callable
import time

def param_timer_deco(func: Callable):
    def wrapper(*args, **kwargs):
        start = time.time()
        res = func(*args, **kwargs)
        end = time.time()
        print(f"Исполнение заняло {end-start}")
        return res
    return wrapper
```

```python
@param_timer_deco
def my_func(sleep_time: int):
    time.sleep(sleep_time)
    return 124

my_func(1)
```

## Декоратор с параметром

```python
from typing import Callable
import time
from functools import wraps


def limit_calls(limit: int):
    def wrapper(func: Callable):
        @wraps(func)
        def inner(*args, **kwargs):
            """Лол привет"""
            nonlocal limit
            if limit == 0:
                print("Нельзя вызвать функцию")
                return

            res = func(*args, **kwargs)
            limit -= 1
            return res
        return inner
    return wrapper
```

```python
@limit_calls(2)
def my_func(sleep_time: int):
    """Очень важный докстинг не потерять его!!!"""
    time.sleep(sleep_time)
    return 124
```

## Декоратор для асинхронных функций

```python
from typing import Coroutine
import asyncio


def deco(coroutine: Coroutine):
    async def wrapper(*args, **kwargs):
        res = await coroutine(*args, **kwargs)
        return res
    return wrapper


@deco
async def my_async_func():
    await asyncio.sleep(0.5)
    return 1

await my_async_func()

# my_async_func = deco(my_async_func)

```

```python
from functools import lru_cache


@lru_cache
def my_long_calc():
    time.sleep(3)
    return 42
```

```python
print(my_long_calc())
print(my_long_calc())
print(my_long_calc())
```
