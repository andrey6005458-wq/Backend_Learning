# **Примеры структур со словарями(dict)**

### **`Список всех ключей - итерация`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
for key in names_marks:
    print(key)
```

### **`Список всех занчений - итерация`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
for val in names_marks.values():
    print(val)
```

### **`Список пары (ключ, значение) - итерация`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
for key, val in names_marks.items():
    print(f'{key}: {val}')
```

### **`Получение значения по ключу`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
print(names_marks['Андрей'])
```

### **`Добавить или обновить ключ`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
names_marks['Алексей'] = 4
for key, val in names_marks.items():
    print(f'{key}: {val}')
```

### **`Удалить ключ`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
del names_marks['Владислав']
print(*names_marks.items(), sep=', ')
```

### **`Проверка на ключ в словаре с помощью get`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
if 'Андрей' in names_marks:
    print(names_marks['Андрей'])
```

### **`Проверка на ключ в словаре с помощью get`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
x = names_marks.get('Антон', 'Такого ключа в словаре нет')
print(x)
```

### **`Длина словаря`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
print(len(names_marks))
```

### **`Сортировка по ключам и значениям в алфавитном порядке`**

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
for students, marks in sorted(names_marks.items()):
    print(f'{students}: {marks}')
```

### **`Проверка на наличие ключа и добавление новой пары ключ - значение`**<br>(*с выводом построчно, сортированно по алфавиту, с выравниванием по левому краю*)

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
names_marks.setdefault('Ирина', 5)
for key, values in sorted(names_marks.items()):
    print(f'{key:<9}: {values}')
```

```python
names_marks = {
    "Андрей": 5,
    "Кристина": 4,
    "Дмитрий": 5,
    "Владислав": 3,
    "Максим": 5
}
print(names_marks)
```
